### python과 데이터베이스 연동
- 내장 DB 연동 : SQLite

```python
import sqlite3

# isolation_level : 트랜잭션
# autocommit 중지
conn = sqlite3.connect("./test.db", isolation_level = None)

# sql을 데이터베이스에 전달하고 결과를 받아오는 객체
cursor = conn.cursor();
```

```python
# 테이블 생성
sql = """create table if not exists users(id integer primary key, username text, email text, phone text, website text, regdate text)"""

cursor.execute(sql)
```

```python
from datetime import datetime
now =  datetime.now()
nowDateTime = now.strftime("%Y-%m-%D %H:%M:%S")
nowDateTime
```

```python
# insert 구문

sql = """insert into users(id, username, email, phone, website, regdate) values(?,?,?,?,?,?)"""

# 튜플형태로 작성하기
cursor.execute(sql, (2, 'park', 'park@naver.com', '010-1234-4444', 'park.com', nowDateTime))
```

```python
user_list = [
    (3, 'YOO', 'yoo@naver.com', '010-5353-4444', 'yoo.com', nowDateTime),
    (4, 'hong', 'hong@naver.com', '010-6666-6666', 'hong.com', nowDateTime),
    (5, 'choi', 'choi@naver.com', '010-2424-6666', 'choi.com', nowDateTime),
]

cursor.executemany(sql, user_list)
```

```python
# 데이터 조회

sql = "select * from users"

cursor.execute(sql)
```

```python
sql = "select * from users"

cursor.execute(sql)

for row in cursor.fetchall():
    print(row)
```

```python
sql = "select * from users where id=3"
cursor.execute(sql)
cursor.fetchall() 
```

```python
sql = "select * from users where id=?"
cursor.execute(sql, (2,))
cursor.fetchall() 
```

```python
# 두개 가져오기

sql = "select * from users where id in (?,?)"
cursor.execute(sql, (3,5))
cursor.fetchall() 
```

```python
sql = "update users set username=? where id=?"
cursor.execute(sql, ('young', 2))
```

```python
cursor.execute("select * from users where id=2")
cursor.fetchone()
```

```python
# 삭제

sql = "delete from users where id=?"
cursor.execute(sql,(5,))
```
