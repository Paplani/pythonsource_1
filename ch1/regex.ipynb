#### 정규표현식(정규식)
- 복잡한 문자열을 처리할 때 사용하는 방법
- 파이썬만의 고유한 문법이 아니라 문자열을 처리하는 모든 곳에서 사용하는 일종의 형식 언어

```python
# 주민등록번호 => 뒷자리를 모두 * 문자로 변경

# 파이썬에서 정규식 지원하는 모듈

import re

dir(re)

# re.compile() : 정규식을 작성
pattern = re.compile("ab*")

# 패턴을 특정 문자열이 포함하고 있는가?       
print(pattern.match("abc"))
```

| 메서드 | 목적 |
|-----|-----|
|`match` | 문자열의 처음부터 정규식과 매치되는지 조사|
|`search` | 문자열 전체를 검색하여 정규식과 매치되는지 조사|
|`findall` | 정규식과 매치되는 모든 문자열(substring)을 리스트로 리턴|
|`finditer` | 정규식과 매치되는 모든 문자열(substring)을 이터레이터 객체로 리턴|

```python
# [문자] : [] 안에 들어있는 문자들과 매치하느냐??

p = re.compile("[abc]")

print(p.match("a"))
print(p.match("before"))
print(p.match("dude"))
```

```python
m = p.match("a")

print(f"패턴 시작 : {m.start()}")
print(f"패턴 끝 : {m.end()}")
print(f"패턴과 일치하는 문자열 반환 : {m.group()}")
print(f"일치의 시작, 끝 반환 : {m.span()}")

if m:
    print(f"Match found {m.group()}")
else:
    print("No Match")
```

```python
p = re.compile("[abc]")
m = p.findall("Happy forever")
print(m)

# finditer를 통해 받으면 객체로 받을수 있음. for문 가능
m = p.finditer("Happy forever")
print(m)

for r in m:
    print(r)
```

```python
# . : 마침표로서 사용하고 싶을 때
# \. or [.] : 이래야 마침표로 인식함
# \t, \r ... : 이스케이프 문자로 인식할수 있기 때문에 앞에 r을 써줘야함

p = re.compile(r"a\.b")
# p = re.compile("a\.b") => 오류뜸

print(p.match("a\nb"))
print(p.match("abb"))
print(p.match("a0b"))
print(p.match("aab"))
print(p.match("a.b"))
```

- 자주 사용하는 문자 표현식

| 정규표현식 | 설명 |
|-----|----|
| \d | 숫자와 매치 [0-9]와 동일한 표현식 |
| \D | 숫자가 아닌 것과 매치 [^9]와 동일한 표현식 |
| \s | 화이트스페이스(스페이스, 탭, 줄바꿈 등)문자와 매치 [빈칸\t\n\r\f\v]와 동일한 표현식 |
| \S | 화이트스페이스(스페이스, 탭, 줄바꿈 등)문자가 아닌 것과 매치 [빈칸\t\n\r\f\v]와 동일한 표현식 |
| \w | 문자 + 숫자와 매치 [a-zA-Z0-9_]와 동일한 표현식 |
| \W | 문자 + 숫자가 아닌 것과 매치 [^a-zA-Z0-9_]와 동일한 표현식 |

```python
p = re.compile("[0-9]+")
p = re.compile(r"\d+")
print(p.match("012 abc"))

# 원본 문자열에서 0~9 가 아닌 문자 찾기
# p = re.compile("[^0~9]+")
p = re.compile(r"\D+")
print(p.match("012 abc"))

# 공배그 태브 엔터 문자 찾기
p = re.compile("[\t\n\r\f\v]+")
p = re.compile(r"\s+")
print(p.search("012 abc"))

# 공백, 탭, 엔터 문자가 아닌 문자 찾기
# p = re.commpile("[^ \t\v\r\f\v]+")
p = re.compile(r"\S+")
print(p.search("012 abc"))

# 대소문자, 숫자, _ 포함한 문자 찾기
# p = re.compile("[a-zA-Z0-9_]+")
p = re.compile(r"\w+")
print(p.search("012 abc"))


# 대소문자, 숫자, _가 아닌 문자 찾기
# p = re.compile("[^a-zA-Z0-9_]+")
p = re.compile(r"\W+")
print(p.search("012 abc"))
```

```python
# re.M : MULTILINE
# re.M 옵션을 주면, 각 줄(line)의 시작마다 ^ 체크를 하겠다는 뜻

data = """python one
life is too short
python two
you need python
python three
"""

# ^ : 시작하는
# python 시작하고 스페이스 문자 하나 영어, 숫자, , _인 문자 찾아줘
p = re.compile(r"^python\s\w+", re.M)
print(p.findall(data))
```

```python
# 아이디, 비밀번호, 전화번호 입력받기
# 패턴에 맞을 때까지 입력받기

# 아이디 검사 조건
# 영문 대문자로 시작, 영문, 숫자, 특수문자(!@#$&) 포함, 6~15 자

# 비밀번호 검사 조건
# 영문 대소문자,숫자, 특수문자(~~~~~) 포함, 8자 이상

# 전화번호 010-1234-5678 / 01012345678
# fullmatch()

import re

user_info = []

while True:
    user_id = input("아이디를 입력해주세요.")
    # A12345678 => 통과됨.
    p = re.compile(r"[A-Z][a-zA-z0-9!@#$&]{5,14}")
    correct = re.fullmatch(p, user_id)
    if correct :
        user_info.append(user_id)
        print("아이디 생성 완료")
        break
    else: 
        print("아이디를 다시 입력해주세요")

while True:
    user_ps = input("비밀번호를 입력해주세요.")
    if (len(user_ps) >= 8 
    and re.search("[A-Z]", user_ps) 
    and re.search("[a-z]", user_ps) 
    and re.search("\d", user_ps)
    and re.search("[!@#$&]", user_ps)
    ):
        print("사용 가능한 비밀번호")
        break
    else:
        print("비밀번호를 다시 입력해주세요")
        

while True:
    user_num = input("전화번호를 입력해주세요.")
    p = re.compile(r"\d{3}-\d{4}-\d{4}$")
    correct = re.fullmatch(p, user_num)
    if correct:
        user_info.append(user_num)
        print("전화번호 입력 완료")
        break
    else:
        print("전화번호를 다시 입력해주세요.")

print(user_info)
```
