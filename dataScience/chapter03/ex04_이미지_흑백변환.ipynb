### 이미지 업로드 + 흑백 변환기
- gr.Image로 이미지를 업로드받고 처리 결과를 다시 이미지로 보여주는 출력
- 텍스트/숫자뿐 아니라 이미지도 Gradio의 입출력이 가능

```python
import gradio as gr
import numpy as np        # 이미지 배열 계산

def to_grayscale(input_image):
    # input_image: [가로, 세로, rgb],   ...은 그대로 두고 전체 가져옴 / 색상표는 투명도 조정없이 3으로 선택
    gray=np.dot(input_image[..., :3], [0.299, 0.587, 0.114])   # 표준공식

    gray_image = np.stack([gray, gray, gray], axis=1).astype(np.unit8)
    return gray_image
```

```python
demo = gr.Interface(
    fn = to_grayscale,
    inputs = gr.Image(label = "원본 이미지 업로드"),
    outputs = gr.Image(label="흑백 변환 결과"),
    title= "이미지 흑백 변환기",
)
demo.launch()
demo.close()
```
